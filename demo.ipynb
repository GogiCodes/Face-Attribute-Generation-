## 3. Model Architecture Overview

Building and inspecting the Conv-VAE model.

```python
# Import model components
sys.path.insert(0, str(project_dir))

from model import ConvVAE, Encoder, Decoder, vae_loss
from config import Config, DEFAULT_CONFIG

# Initialize model
model = ConvVAE(
    latent_dim=64,
    image_channels=3
)

# Move to device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)

# Count parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print("🧠 CONV-VAE MODEL ARCHITECTURE")
print("=" * 50)
print(f"\n{'Component':<25} {'Parameters':>15}")
print("-" * 50)
print(f"{'Encoder':<25} {sum(p.numel() for p in model.encoder.parameters()):>15,}")
print(f"{'Decoder':<25} {sum(p.numel() for p in model.decoder.parameters()):>15,}")
print("-" * 50)
print(f"{'TOTAL':<25} {total_params:>15,}")
print(f"{'Trainable':<25} {trainable_params:>15,}")

print(f"\n✓ Model initialized on device: {device}")
print(f"✓ Latent dimension: 64")
print(f"✓ Input resolution: 64×64×3 (RGB)")
```

## 4. Model Capabilities Demo

Demonstrating key model capabilities:
- Random face generation from prior N(0, I)
- Image reconstruction through encoding-decoding
- Latent space properties

```python
print("🎨 MODEL CAPABILITIES DEMONSTRATION")
print("=" * 50)

# 1. Generate random samples
print("\n1️⃣  Generating random faces from N(0, I)...")
with torch.no_grad():
    z_random = torch.randn(4, 64, device=device)
    generated_faces = model.decode(z_random)
print(f"   ✓ Generated 4 random faces: {generated_faces.shape}")
print(f"   Pixel range: [{generated_faces.min():.3f}, {generated_faces.max():.3f}]")

# 2. Test reconstruction
print("\n2️⃣  Testing encoding-decoding cycle...")
with torch.no_grad():
    dummy_image = torch.rand(2, 3, 64, 64, device=device)
    mu, logvar = model.encoder(dummy_image)
    z = model.reparameterize(mu, logvar)
    reconstructed = model.decode(z)
print(f"   ✓ Input shape: {dummy_image.shape}")
print(f"   ✓ Latent code shape: {z.shape}")
print(f"   ✓ Reconstruction shape: {reconstructed.shape}")
print(f"   ✓ Reconstruction MSE: {torch.mean((reconstructed - dummy_image) ** 2):.4f}")

# 3. Latent space properties
print("\n3️⃣  Analyzing latent space properties...")
with torch.no_grad():
    sample_size = 100
    latent_codes = torch.randn(sample_size, 64, device=device)
    print(f"   ✓ Sampled {sample_size} latent vectors")
    print(f"   Mean: {latent_codes.mean(dim=0)[:5].cpu().numpy()}... (showing first 5)")
    print(f"   Std:  {latent_codes.std(dim=0)[:5].cpu().numpy()}... (showing first 5)")
    print(f"   All means close to 0: {torch.allclose(latent_codes.mean(dim=0), torch.zeros(64, device=device), atol=0.3)}")

print("\n✅ All capability tests passed!")
```
